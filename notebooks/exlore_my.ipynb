```python
tables = [df_order, df_detail, df_pizza, df_type]
table_names=['df_order', 'df_detail', 'df_pizza', 'df_type']
```

```python
for table in tables:
    print(table.shape)
```

```python
print(f"Description des variables __numériques__")
print("\n")
for i in range(len(tables)):
    print(f" Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].describe().T)
    print("end================================")
    print("\n")
```

```python
print("Déscription des variables __categoriels__")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].describe(include=['string','category']).T)
    print("end================================")
    print("\n")
```

```python
print("recherche des valeurs manquantes dans les tables")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].isnull().sum())
    print("end================================")
    print("\n")
```

```python
print("Recherche des doublons dans les tables")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].duplicated().sum())
    print("end================================")
    print("\n")
```
