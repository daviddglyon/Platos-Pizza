```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_order = pd.read_csv('../data/orders.csv')
df_detail = pd.read_csv('../data/order_details.csv')
df_pizza = pd.read_csv('../data/pizzas.csv')
```

```python
df_type = pd.read_csv('../data/pizza_types.csv', encoding='latin-1')
```

```python
tables = [df_order, df_detail, df_pizza, df_type]
table_names=['df_order', 'df_detail', 'df_pizza', 'df_type']
```

```python
print(f"Les tailles des tables sont :")
print("\n")
for i in range(len(tables)):
    print(f" Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].shape)
    print("end================================")
    print("\n")
```

```python

print(f"Description des variables __numériques__")
print("\n")
for i in range(len(tables)):
    print(f" Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].describe().T)
    print("end================================")
    print("\n")
```

```python

print("Déscription des variables __categoriels__")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].describe(include=['string','category']).T)
    print("end================================")
    print("\n")
```

```python
print("Recherche des valeurs manquantes dans les tables")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].isnull().sum())
    print("end================================")
    print("\n")
```

```python
print("Recherche des doublons dans les tables")
print("\n")
for i in range(len(tables)):
    print(f"Table: <<<{table_names[i]}>>>:")
    print("--------------------------------")
    print(tables[i].duplicated().sum())
    print("end================================")
    print("\n")
```

```python
# on remplace le symbole "" par une apostrophe simple dans la colonne ingredients de la table df_type
df_type['ingredients'] = df_type['ingredients'].str.replace("", "'", regex=False)
df_type.query('ingredients.str.contains("")')
```
